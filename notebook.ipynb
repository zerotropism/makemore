```python
# declare bigram dictionnary
b = {}
# consecutive elements in the first word
for w in words:
    # adding "start" & "end" characters wrapped into a list of the chars composing the word
    chs = ["<S>"] + list(w) + ["<E>"]
    for ch1, ch2 in zip(chs, chs[1:]):
        bigram = (ch1, ch2)
        b[bigram] = b.get(bigram, 0) + 1
```

```python
# showing the bigram dictionnary sorted by decreasing occurence
sorted(b.items(), key=lambda kv: -kv[1])
```
